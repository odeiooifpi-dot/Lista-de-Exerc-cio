Um estacionamento cobra R$ 8,00 por até 2 horas. Acima de 2 e até 5 horas, cobra R$ 8,00 mais.
R$ 3,00 por hora adicional. Acima de 5 horas, cobra R$ 17,00 mais R$ 2,00 por hora que
ultrapassar 5. Leia a quantidade inteira de horas e calcule o valor a pagar.

```python
h = int(input("Horas: "))
if h <= 2:
    valor = 8
elif h <= 5:
    valor = 8 + (h - 2) * 3
else:
    valor = 17 + (h - 5) * 2
print(f"Valor a pagar: R$ {valor:.2f}")
```

Um ingresso custa R$ 25,00. Solicite valores inteiros ao usuário até que o total inserido seja
suficiente para a compra. Ao final, informe o total inserido e o troco.

```python
total = 0

while total < 25:
    total += int(input("Digite um valor: "))

print(f"Total inserido: R$ {total:.2f}")
print(f"Troco: R$ {total - 25:.2f}")
```

Leia 8 números inteiros. Informe o maior valor e a posição em que ele apareceu pela primeira
vez.

Leia 8 números inteiros distintos e informe o maior e o segundo maior. Não utilize max(), sort() ou listas.

```python
maior = segundo = None

for i in range(8):
    n = int(input("Digite um número: "))
    if maior is None or n > maior:
        segundo = maior
        maior = n
    elif segundo is None or n > segundo:
        segundo = n

print("Maior:", maior)
print("Segundo maior:", segundo)
```

O programa começa com saldo de R$ 500,00. Serão realizadas 6 operações. Em cada uma, leia
o tipo (1 = depósito; 2 = saque) e o valor. Depósitos aumentam o saldo. Saques só podem
ocorrer se houver saldo suficiente; caso contrário, mostre 'Saldo insuficiente'. Ao final, mostre o
saldo.

```python
saldo = 500

for i in range(6):
    tipo = int(input("Tipo 1 depósito ou 2 saque: "))
    valor = float(input("Valor: "))

    if tipo == 1:
        saldo += valor
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")

print(f"Saldo: R$ {saldo:.2f}")
```

Leia um valor inteiro positivo e informe a menor quantidade de cédulas para representá-lo,
usando R$ 100, 50, 20, 10, 5, 2 e 1.

```python
valor = int(input("Digite o valor: "))

for cedula in [100, 50, 20, 10, 5, 2, 1]:
    qtd = valor // cedula
    valor %= cedula
    print(f"R$ {cedula}: {qtd}")
```

Leia um inteiro maior que 1 e informe se ele é primo.

```python
n = int(input("Digite um número: "))
primo = True

for i in range(2, n):
    if n % i == 0:
        primo = False
        break

print("é primo" if primo else "Não é primo")
```

Leia um inteiro positivo N, mostre todos os seus divisores positivos e informe quantos divisores
ele possui.

```python
n = int(input("Digite um número: "))
qtd = 0

for i in range(1, n + 1):
    if n % i == 0:
        print(i)
        qtd += 1

print("Quantidade de divisores:", qtd)
```

Leia A e B positivos, com A < B, mostre os primos entre A e B (inclusive) e informe quantos foram
encontrados.

```python
A = int(input("A: "))
B = int(input("B: "))
qtd = 0

for n in range(A, B + 1):
    primo = n > 1
    for i in range(2, n):
        if n % i == 0:
            primo = False
            break
    if primo:
        print(n)
        qtd += 1

print("Quantidade:", qtd)
```

Considere o número secreto 37. O usuário terá no máximo 5 tentativas. Após cada erro, informe
se o secreto é maior ou menor que o palpite. Termine ao acertar ou ao usar as 5 tentativas.

```python
secreto = 37

for i in range(5):
    palpite = int(input("Palpite: "))

    if palpite == secreto:
        print("Acertou!")
        break
    elif palpite < secreto:
        print("O secreto é maior")
    else:
        print("O secreto é menor")
```
